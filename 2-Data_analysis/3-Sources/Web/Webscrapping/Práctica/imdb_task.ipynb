```python
from bs4 import BeautifulSoup as bs
import requests
import pandas as pd
```

```python
url = "https://www.imdb.com/chart/top/"
response = requests.get(url)
print (response)
```

```python
from fake_useragent import UserAgent
ua = UserAgent()
headers = {'User-Agent': ua.random}
# Configuración importante para evitar bloqueos
'''
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36',
    'Accept-Language': 'es-ES,en;q=0.9'
}
'''
response = requests.get(url, headers=headers)
htlm = response.text
response
```

```python
# Analizar el contenido HTML de la página
soup = bs(htlm,'html.parser')
```

```python
soup.find_all('h3', class_='ipc-title__text')[1:-1]
```

```python
list_ranking = []
list_nombre = []

for x in soup.find_all('h3', class_='ipc-title__text')[1:-1]:
    list_ranking.append (x.get_text().split(". ")[0]) 
    list_nombre.append (x.get_text().split(". ")[1])# Separa en dos columnas
```

```python
print (list_ranking)
```

```python
soup.find_all('div', class_ = 'sc-4b408797-7 fUdAcX cli-title-metadata')
```

```python
list_anio = []
list_duracion = []
list_edad = []

for x in soup.find_all('span', class_ = 'sc-4b408797-8 iurwGb cli-title-metadata-item'):

    if len(x.text) == 4:
        list_anio.append (x.text)
    if len(x.text) > 4:
        list_duracion.append (x.text)
    if len(x.text) > 0 and len(x.text) <= 2:
        list_edad.append (x.text)
```

```python
print (list_anio)
print (list_duracion)
print (list_edad)
```

```python
list_rating =[]

for x in soup.find_all('span', class_= 'ipc-rating-star--rating'):
    list_rating.append (x.text)
```

```python
print (list_rating)
```

```python
df = pd.DataFrame({
    'Ranking': list_ranking,
    'Nombre': list_nombre,
    'Año': list_anio,
    'Duración': list_duracion,
    'Edad': list_edad
})
```

```python
df
```

```python
df.to_csv('top250_imdb_actualizado.csv', index=False)
print(df)
```
