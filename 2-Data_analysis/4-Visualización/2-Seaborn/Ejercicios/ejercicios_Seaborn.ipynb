```python
# 19. Veamos la función melt() de pandas. Melt toma los valores de varias columnas y los junta todos en una misma columna nueva.
# Dado un DataFrame, melt recibe varios parámetros:
# * qué columnas del DataFrame se quedan igual
# * qué nombres de columnas del dataframe pasan a ser una columna nueva
# * cómo queremos llamar esas nuevas columnas

# Primero recuerda el Dataframe df_grass_fire_water con las primeras 5 filas
```

```python
df_12.head()
```

```python
# 20. Queremos dejar igual las variables 'Name' y 'Type 1'
# Queremos añadir una columa nueva cuyos valores sean las columnas que queremos apilar y que se llame 'New Column'
# Escribimos df_gfw.melt(id_vars=['Name', 'Type 1'], value_vars=[...], var_name='New Column')
```

```python
# Mantenemos 'Name' y 'Type 1', juntamos el resto de 
df_melted = pd.melt(
    df_12,
    id_vars=["Name", "Type 1"],           # columnas que mantenemos fijas
    value_vars=["HP", "Attack", "Defense", "Sp. Atk", "Sp. Def", "Speed"],  # columnas que juntamos
    var_name="New Column",                      # nueva columna con los nombres de las estadísticas
    value_name="Valor"                    # nueva columna con los valores
)
```

```python
# 21. Pintamos el swarmplot con el melted y entendemos la información
```

```python
sns.swarmplot(x="New Column", y="Valor", hue="Type 1", data=df_melted)

plt.title("Distribución por Type 1")
plt.legend(title="Type 1");
```

```python
# 22. Basándote en el gráfico anterior, define el tamaño de la figura en (10,6) (usa plt)
# En el swarmplot añade el argumento dodge a True y usa la anterior paleta de colores Pokemon

# Dodge en inglés es "esquivar", ¿qué diferencia ves?

# Usando plt 
# fija el límite del eje Y entre 0 y 200
# fija la leyenda con bbox_to_anchor en upper left. Pista: upper left es el código 2 y se usa el parámetro loc.
```

```python
plt.figure(figsize=(10, 6))

sns.swarmplot(x="New Column", y="Valor", hue="Type 1", data=df_melted, dodge = True, palette = colores)

plt.title("Distribución por Type 1")
plt.ylim(0,200)
plt.legend(title="Type 1")
plt.legend(title="Type 1", loc=2, bbox_to_anchor=(0, 1));
```

```python
# Observa que incluso usando Seaborn, usamos Pyplot para trabajar con los ejes, la leyenda y el tamaño
# Lo importante es que Seaborn y Pyplot conviven bien juntos
```

```python
# 26. Usa catplot con el DataFrame original. Muestra el valor de Defense según el 'Type 1', 
# muestra un subplot para legendarios y otro para no legendarios
# y divide los colores según el Type 2
```

```python
g = sns.catplot(
    data=df,
    x="Type 1",
    y="Defense",
    hue="Type 2",
    col="Legendary",         
    kind="strip",            
    height=5,
    aspect=1.2
)

# Ajustar etiquetas y diseño
g.set_xticklabels(rotation=45)
g.set_axis_labels("Type 1", "Defense")
g.fig.subplots_adjust(top=0.85)
g.fig.suptitle("Defense según Type 1, separado por Legendary y coloreado por Type 2");
```

```python
# 27. Muestra una función distribución de probabilidad con kdeplot (curvas de nivel) de la defensa respecto al ataque.
# Di un bin 2D donde caigan muchos puntos (ataque, defensa).
```

```python
plt.figure(figsize=(10,6))

sns.kdeplot(
    data=df,
    x="Attack",
    y="Defense",
    fill=True,       # rellena la densidad
    cmap="mako",     # mapa de colores
    thresh=0,        # muestra todos los niveles
    levels=20,       # número de curvas de nivel
    bw_adjust=0.8    # suavidad (ajustar si quieres más o menos detalle)
)

plt.title("Distribución de probabilidad conjunta de Attack vs Defense")
plt.xlabel("Attack")
plt.ylabel("Defense");
```

```python
# 28. Pinta una distribución de probabilidad conjunta con jointplot, 
# pintando la defensa respecto al ataque del DataFrame original
```

```python
plt.figure(figsize=(10,6))

sns.jointplot(
    data=df,
    x="Attack",
    y="Defense",
    kind="kde",
    fill=True,
    cmap="mako",
    thresh=0,
    levels=20,
    bw_adjust=0.8
)

plt.suptitle("Distribución conjunta de Attack y Defense", y=1.02);
```
