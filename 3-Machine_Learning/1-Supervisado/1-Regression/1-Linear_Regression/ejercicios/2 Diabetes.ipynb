## Building a regression model

The data have already been split to predictor (*data*) and response (*target*) variables. Given this information, we'll apply what we have previously learned about linear regression and apply the algorithm to the diabetes dataset.

#### In the cell below, import the `linear_model` class from `sklearn`. 

```python
cmap = sns.diverging_palette(10, 240, sep=80, as_cmap=True)
sns.heatmap(diabetes_df.corr(numeric_only=True), annot=True, cmap=cmap, center=0, vmin=-1, vmax=1, linewidths=0.5, linecolor="gray");
```

#### Create a new instance of the linear regression model and assign the new instance to the variable `diabetes_model`.

```python
# Cojo estas variables porque son los que tienen las correlaciones más altas con el target

X = diabetes_df[['bmi', 'bp', 's3','s4', 's5', 's6']]
y = diabetes_df['target']
```

#### Next, let's split the training and test data.

Define `diabetes_data_train`, `diabetes_target_train`, `diabetes_data_test`, and `diabetes_target_test`. Use the last 20 records for the test data and the rest for the training data.

```python
from sklearn.model_selection import train_test_split

diabetes_data_train, diabetes_data_test, diabetes_target_train, diabetes_target_test = train_test_split(X, y, test_size = 20, shuffle= False)
```

```python
print('Tamaño X train', diabetes_data_train.shape)
print('Tamaño X test', diabetes_data_test.shape)
print('Tamaño y train', diabetes_target_train.shape)
print('Tamaño y test', diabetes_target_test.shape)
```

Perform a brief exploratory analysis to investigate the conditions necessary for linear regression. This analysis will help you understand the distribution of the target variable and its correlations with other features.

```python
sns.displot(diabetes_df['target']);
```

Fit the training data and target to `diabetes_model`. Print the *intercept* and *coefficients* of the model.

```python
from sklearn.linear_model import LinearRegression
lm = LinearRegression()
```

```python
lm.fit(diabetes_data_train, diabetes_target_train)
```

```python
print('Intercepto', lm.intercept_)
print('Coeficiente', lm.coef_)
```

#### Inspecting the results

From the outputs you should have seen:

- The intercept is a float number.
- The coefficients are an array containing 10 float numbers.

This is the linear regression model fitted to your training dataset.

#### Using your fitted linear regression model, predict the *y* of `diabetes_data_test`.

y = a + b*x1 + c*x2 + d*x3 + ....

```python
predicciones = lm.predict (diabetes_data_test) 
predicciones
```

#### Print your `diabetes_target_test` and compare with the prediction. 

```python
df_pred = pd.DataFrame({'pred': predicciones, 'valor_real': diabetes_target_test})
df_pred ['Error'] = df_pred ['valor_real'] - df_pred ['pred']
df_pred
```

```python
sns.scatterplot(x = diabetes_target_test, y = predicciones);
```

#### Is `diabetes_target_test` exactly the same as the model prediction? Use different error metrics

```python
sns.displot((diabetes_target_test - predicciones));
```

```python
from sklearn import metrics
 
```

```python
print('MAE:', round(metrics.mean_absolute_error(diabetes_target_test, predicciones),3))
print('MAPE', round(metrics.mean_absolute_percentage_error(diabetes_target_test, predicciones),3))
print('MSE:', round(metrics.mean_squared_error(diabetes_target_test, predicciones),3))
print('RMSE:', round(np.sqrt(metrics.mean_squared_error(diabetes_target_test, predicciones)),3))
print('score train',round(lm.score(diabetes_data_train, diabetes_target_train),3))
print('score test',round(lm.score(diabetes_data_test, diabetes_target_test),3))      
```
